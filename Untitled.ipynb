```python
from scipy.sparse.linalg import minres
import numpy as np
```

```python
A = np.array([[2,3],[1,5]])
b = np.array([3,4])
x , info = minres(A,b)
```

```python
np.random.seed(10000)
n = 10000
M = np.eye(n)
B = np.random.rand(n,n//2)
A = np.block([[M,B],[B.T,np.zeros((n//2,n//2))]])
b = np.random.randn(A.shape[0])
```

```python
x , exitcode = minres(A,b,rtol=1e-05,maxiter=500)
```
