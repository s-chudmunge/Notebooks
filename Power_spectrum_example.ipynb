```python
import math
import torch
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.linspace(0,2*math.pi)
function = np.cos(2*x)+np.exp(np.sin(4*x))
```

```python
plt.plot(x,function,'--')
plt.xlabel("x")
plt.ylabel("f(x)")
plt.show()
```

```python
N = 1024
L = 2 * math.pi
x = np.linspace(0, L, N, endpoint=False)
f = np.cos(2*x) + np.exp(np.sin(4*x))

# FFT
fk = np.fft.fft(f)

# power spectrum
Pk = np.abs(fk)**2 / N

# wavenumbers
k = np.fft.fftfreq(N, d=L/N)

# keep only positive k
mask = k > 0
k = k[mask]
Pk = Pk[mask]

# plot
plt.plot(k, Pk)
plt.xlabel("k")
plt.ylabel("P(k)")
plt.yscale("log")
plt.show()
```
