```python
import numpy as NP
import matplotlib.pyplot as PP
```

```python
phi = NP.loadtxt('phi.dat')
```

```python
NP.shape(phi)
```

```python
PP.imshow(phi)#cmap='jet',origin='lower',extent=[0. ,1., 0., 1.])
PP.colorbar()
```

```python
PP.contourf(phi,20,cmap='jet')
PP.colorbar()
```

```python
x = NP.linspace(0,1,99)
y = NP.linspace(0,1,99)
X,Y = NP.meshgrid(x,y)
fig, ax = PP.subplots(subplot_kw={"projection": "3d"})

ax.plot_surface(X,Y,phi,cmap = 'jet')
```
